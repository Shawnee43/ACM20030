## Q1 a

Write a function that checks if a given $n\times n$ matrix is symmetric. Use only loops and `.shape` function from numpy (do not use any other special numpy functions).

```python
def check(A):
    count = 0
    rows = np.shape(A)[0]
    columns = np.shape(A)[1]
    for i in range(rows):
        for j in range(columns):
            if A[i, j] != A[j, i]:
                count += 1
    
    if count == 0:
        print('matrix is symmetric')
    else:
        print('matrix not symmetric')
        
```

## Q1 b

Write a function to transpose a general $n\times n$ matrix. Do this by writing loops and not using any special functions like `np.transpose()`, `A.T`, or any functions from any other library (`np.zeroes`, `np.ones`, `np.shape`, etc are all ok to use).

```python
def transpose(A):
    rows = np.shape(A)[0]
    columns = np.shape(A)[1]
    T = np.zeros((columns, rows))
    for i in range(rows):
        for j in range(columns):
            T[j, i] = A[i, j]
            
    return T
```

## Q1c

Write a function that finds the maximum entry in a matrix (where the matrix does not need to be square). Do this by writing loops and not using any special functions like `np.max` or any functions from any other library.

```python
def max_array(A):
    rows = np.shape(A)[0]
    columns = np.shape(A)[1]

    current_max = A[0][0]
    
    for i in range(rows):
        for j in range(columns):
            if A[i][j] > current_max:
                current_max = A[i][j]
    return current_max
  
```

Check your result by comparing to `np.max(A)` for the below random matrix

```python
A = np.random.rand(4,5)
print(np.max(A))
print(max_array(A))
```

# Q2a

The $L^1$ norm is given by

$$ \lVert A \rVert_1 = \max_j \sum^m_{i=1} |a_{ij}| $$

i.e., it is the max of the sums of the absolute values of each column. Write a function that computes the $L^1$-norm using only loops. The function should use no special numpy functions except `np.abs()` (`np.shape`, `np.inf`, etc are ok to use).

```python
def L1_norm(A):
    
    for j in range(len(A[0])):
        columns = np.abs(A[:, j])      

        sum_ = 0
        for i in range(len(A[:, 0])):
            sum_ += columns[i]         

       
        if j == 0:
            max_sum = sum_
    
        elif sum_ > max_sum:
            max_sum = sum_

    return max_sum   
       
    
```

Check your result against NumPy's `la.norm(A,1)` function for the following random matrix

```python
A = np.random.rand(3,4)
print(la.norm(A, 1))
print(L1_norm(A))
```

## Q2b

The $L^\infty$ norm is given by 

$$\lVert A \rVert_\infty = \max_i \sum_{j=1}^n |a_{ij}|$$

i.e., it is the max of the sums of the absolute values of each row. Write a function that computes the $L^\infty$-norm using only loops and no special numpy functions.

```python
def L_inf_norm(A):

    for j in range(len(A)):     
        rows = np.abs(A[j, :])     

        sum_ = 0
        for i in range(len(A[0])): 
            sum_ += rows[i]

        if j == 0:
            max_sum = sum_
        elif sum_ > max_sum:
            max_sum = sum_

    return max_sum 
```

Check your result against NumPy's `la.norm(A,np.inf)` function for the following random matrix

```python
A = np.random.rand(3,4)
print(la.norm(A, np.inf))
print(L_inf_norm(A))

```

$$100\times\frac{\lVert \delta x_i\rVert_2}{\lVert x_i \rVert_2}\quad\text{and}\quad 100\times\frac{\lVert \delta b\rVert_2}{\lVert b \rVert_2}$$

where $(x_i+\delta x_i)  = A_i^{-1}(b+\delta b)$ and $i\in\{1,2\}$.

```python
b = np.array([1, 2, 3, 4])
delta_b = np.array([0, 0.1, 0, 0])

#A1
#matrix xi
A1_inv = la.inv(A1)
xi = A1_inv @ b

#matrix xi+1
A1_inv = la.inv(A1)
xi1 = A1_inv @ (b + delta_b)

delta_x = xi - xi1

#percentage change in x and b
def delta_x_percent(xi, delta_x):
    return 100*(la.norm(delta_x)/la.norm(xi))

def delta_b_percent(b, delta_b):
    return 100*(la.norm(delta_b)/la.norm(b))

#results for A1
print("A1: % change in x =", delta_x_percent(xi, delta_x))
print("A1: % change in b =", delta_b_percent(b, delta_b))

#matrix A2
#matrix xi
A2_inv = la.inv(A2)
xi = A2_inv @ b

#matrix xi+1
A2_inv = la.inv(A2)
xi1 = A2_inv @ (b + delta_b)

delta_x = xi - xi1

#results for A2
print("A2: % change in x =", delta_x_percent(xi, delta_x))
print("A2: % change in b =", delta_b_percent(b, delta_b))
```

Find the eigenvalues and eigenvectors for the below matrix. Print out the eigenvalues sorted by descending value and explicitly show that eigenvalues/eigenvector pairs satisfy $Ax = \lambda x$.

```python
A = np.array([[1,4,5],[4,5,4],[6,7,9]])
```

```python
print(la.eig(A))

def eigenvals(A):
    eigen_vals = la.eig(A)[0]
    eigen_vals = np.sort(eigen_vals)[::-1]
    return eigen_vals

print(eigenvals(A))

#assign corresponding evalues and evectors
evals = la.eig(A)[0]
evectors = la.eig(A)[1]

idx = np.argsort(evals)[::-1]     
evectors = evectors[:, idx] 
evals = evals[idx]

evalue1 = evals[0:1]
evector1 = evectors[:,0]

evalue2 = evals[1:2]
evector2 = evectors[:,1]

evalue3 = evals[2:3]
evector3 = evectors[:,2]

#check evalue equation satisfied
def pair1(A, evalue1, evector1):
    if np.allclose(A.dot(evector1), evalue1 * evector1):    
        print('Eigenvalue equation satisfied')
    else:
        print('not satisfied')

def pair2(A, evalue2, evector2):
    if np.allclose(A.dot(evector2), evalue2 * evector2):    
        print('Eigenvalue equation satisfied')
    else:
        print('not satisfied')

def pair3(A, evalue3, evector3):
    if np.allclose(A.dot(evector3), evalue3 * evector3):    
        print('Eigenvalue equation satisfied')
    else:
        print('not satisfied')

print(pair1(A, evalue1, evector1))
print(pair2(A, evalue2, evector2))
print(pair3(A, evalue3, evector3))

```

# Q4b

Without using NumPy's `la.eig()` (or similar functions), find the 4 eigenvalues and eigenvectors of the following matrix. Hint: Use the power method for the max eigenvalue (given below in `MaxEigenvalue`), the power method applied to the inverse for the smallest eigenvalue, and apply Hotelling's deflation twice to find the second and third eigenvalues/eigenvectors.

```python
A = np.array([[2.65287619, 1.58615494, 1.47170961, 1.40586081],
       [1.58615494, 1.17577323, 0.72084593, 0.79382861],
       [1.47170961, 0.72084593, 1.18111969, 0.6654689 ],
       [1.40586081, 0.79382861, 0.6654689 , 0.84502394]])
```

```python
def MaxEigenvalue(A, err):
    (m,n) = A.shape
    if(m != n):
        print("Matrix must be square")
        return
    
    # Create a random initial vector
    x = np.random.rand(m)
    
    lam = 0.1
    lamprev = 1
    while np.abs(1-lam/lamprev) > err:
        Ax      = A@x
        lamprev = lam
        lam = la.norm(Ax,2)/la.norm(x,2)
        x = Ax
        
    x = x/la.norm(x)
    return (lam,x)
print('The max eigenvalue followed by the eigenvectors are',MaxEigenvalue(A, 1e-6))
```

```python
#Using power method where the smallest eiganvalue is the inverse of the largest eigenvalue
lmin = 1/MaxEigenvalue(la.inv(A), 1e-14)[0]

#extracting correpsonding eigenvectors
evectors_min = 1/MaxEigenvalue(la.inv(A), 1e-14)[1]

print(lmin)
print(evectors_min)
```
