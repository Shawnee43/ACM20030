```python
#Question2(e)
def IsPrime(n):
    if n == 1:
        return 0
    else:
        i = 2
        while i <= np.sqrt(n):
            #for each i check if it divides n
            if (n % i == 0):
                return 0
            i += 1
            #if no divisors are found number is prime
    return 1

count = 0
for i in range(1, 1000):
    if IsPrime(i)== 1:
        count += 1
print("The number of prime numbers less than 1000 is,", count)

#Question3(f)
def PrimeCount(n):
    count2 = 0
    for i in range(1,n):
        count2 += IsPrime(i)
    return count2
    
#two different ways of writing the same thing 

print("The number of prime numbers less than 1000 is,", PrimeCount(1000))
print("The number of prime numbers less than 10000 is,", PrimeCount(10000)) 
```

```python
x = np.arange(0,40,1)
Pi = np.zeros(40)
i = 0 
while i < np.size(Pi):
    Pi[i] = PrimeCount(i)
    i += 1

plt.plot(x, Pi, 'ro')
plt.xlabel('x')
plt.ylabel('Pi(x)')
plt.grid()
```

```python
#Question3
z = np.loadtxt("Data.txt")
print(z)
x = z[:, 0]
y = z[:, 1]

ylog = np.log(y)
plt.plot(x, ylog)
plt.xlabel("x")
plt.ylabel("Log(y)")
plt.title("Log Scale")
plt.grid()

xlog = np.log(x)
plt.figure()
plt.plot(xlog, ylog)
plt.xlabel("Log(x)")
plt.ylabel("Log(y)")
plt.title("Log-Log Scale")
plt.grid()

#Data is representation of the power law as it bends on semi-log plot and is straight
#line on log-log plot. This is because as you take the logs of both axis' you "compress"
#them which essentially gets rid of any curvature.

#n=2






```
