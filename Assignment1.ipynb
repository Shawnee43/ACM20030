```python
#Question2 
def IsPrime(n):
    if n == 1:
        return print("if the number is 1, it's neither prime nor composite") 
    i = 2
    while i <= np.sqrt(n):
        #for each i check if it divides n
        #needs an equal sign to check upto 3
        print(i)
        if (n % i == 0):
            return 0
        i += 1
        #if no divisors are found number is prime
    return 1
print(IsPrime(9))

count = 0
for i in range(1000):
    if IsPrime(i):
        count += 1
print("The number of prime numbers less than 1000 is,", count)


def PrimeCount(n):
    count2 = 0
    for i in range(n):
        if IsPrime(i):
            count2 += 1
        return count

print("The number of prime numbers less than 1000 is,", PrimeCount(1000))
print("The number of prime numbers less than 10000 is,", PrimeCount(10000)) #Fix code

x = np.arange(0,40,1)
Pi = np.zeros(40)
i = 0 
while i < np.size(Pi):
    Pi[i] = PrimeCount(i)
    i += 1

plt.plot(x, Pi, 'ro')
plt.xlabel('x')
plt.ylabel('Pi(x)')
plt.grid()



    
```
